3.Numerical Operators

```r
x <- 5
y <-3

add <- x + y
sub <- x - y
mul <- x * y
div <- x / y
exp <- x ^ y

print(add)
print(sub)
print(mul)
print(div)
print(exp)

cat("+", add)
cat("-", sub)
cat("*", mul)
cat("/", div)
cat("^", exp)
```

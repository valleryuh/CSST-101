12. Add more rules

```r
student_risk <- function(attendance, grade){

if (attendance <75 & grade <75){
  return("High Risk")
}else if (attendance <85 | grade <80){
  return("Moderate Risk")
}else if(attendance >= 85 & grade >= 80){
  return("Low Risk")
}else{
  return("Undefined Risk") # Handle any unclassified cases to ensure a string is always returned
}

}
```

12. Apply the Rule to Mutiple Students

```r
students$Risk <- mapply(
  student_risk,
  students$Attendance,
  students$Grade
)

print(students)
```

16. Mini KRR System/Final System

```r
students <- data.frame(
  student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95, 72, 88, 79, 68),
  Grade = c(90, 70, 85, 77, 65)
)

student_risk <- function(attendance, grade) {

  if (attendance < 75 & grade < 75) {
    return("High risk")

  } else if (attendance < 85 | grade < 80) {
    return("Moderate risk")

  } else {
    return("Low risk")
  }
}

recommend_risk <- function(risk) {

  if (risk == "High risk") {
    return("Immediate academic intervention")

  } else if (risk == "Moderate risk") {
    return("Academic monitoring")

  } else {
    return("Continue regular monitoring")
  }
}

mini_krr <- function(student_id, attendance, grade) {

  risk <- student_risk(attendance, grade)
  recommendation <- recommend_risk(risk)

  cat("Student ID:", student_id, "\n")
  cat("Attendance:", attendance, "%\n")
  cat("Grade:", grade, "\n")
  cat("\nRisk Level:", risk, "\n")
  cat("\nRecommendation:\n", recommendation, "\n")
}

mini_krr("S002", 72, 70)
```
